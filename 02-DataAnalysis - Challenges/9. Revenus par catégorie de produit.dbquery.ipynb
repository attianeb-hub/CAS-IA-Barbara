```sql
--Revenus par catégorie de produit :
--Calculez les revenus totaux générés
--pour chaque catégorie de produit.
--(Tables : SalesOrderHeader,SalesOrderDetail, Product,ProductCategory) -

select
cat.Name as Categorie,
sum(det.LineTotal) as Revenus 
from
atb_adventureworks.saleslt.salesorderheader sal,
atb_adventureworks.saleslt.salesorderdetail det,
atb_adventureworks.saleslt.product prd,
atb_adventureworks.saleslt.productcategory cat
where
sal.SalesOrderID = det.SalesOrderID
and det.ProductID = prd.ProductID
and prd.ProductCategoryID = cat.ProductCategoryID
group by cat.Name
order by Revenus desc;
```
