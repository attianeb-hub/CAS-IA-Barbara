```sql
--Commandes par région :
--Trouvez le nombre de commandes
--pour chaque région (StateProvince),
--triées par région.
--(Tables : Address, SalesOrderHeader) -

select 
adr.StateProvince as Region,
count(sal.SalesOrderID) as NombreCommandes
from 
atb_adventureworks.saleslt.salesorderheader sal,
atb_adventureworks.saleslt.address adr
where
sal.ShipToAddressID = adr.AddressID
group by Region
order by Region;

select * from atb_adventureworks.saleslt.address;

select * from atb_adventureworks.saleslt.salesorderheader;
```
