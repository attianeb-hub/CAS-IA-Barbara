```sql
--Liste des produits :
--Récupérez une liste de tous les
--produits avec leurs noms, prix
--(ListPrice) et catégories.
--(Tables : Product, ProductCategory)

-- Test 23.01.2026
select * from barbara_lakehouse.bronze.product;


select --distinct
prd.ProductID, 
prd.Name as NomProduit, 
prd.ListPrice as PrixProduit,
cat.Name as CategorieProduit 
from 
atb_adventureworks.saleslt.product prd,
atb_adventureworks.saleslt.productcategory cat 
where 
prd.ProductCategoryID = cat.ProductCategoryID
order by 1
```
