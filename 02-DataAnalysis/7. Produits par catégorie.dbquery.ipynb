```sql
--Produits par catégorie :
--Affichez le nombre total de produits
--dans chaque catégorie de produit.
--(Tables : Product, ProductCategory)

select 
cat.Name as CategorieProduit,
count(prd.ProductID) as NombreTotalProduit
from 
atb_adventureworks.saleslt.product prd,
atb_adventureworks.saleslt.productcategory cat 
where 
prd.ProductCategoryID = cat.ProductCategoryID
group by cat.Name
order by count(prd.ProductID) desc;
```
