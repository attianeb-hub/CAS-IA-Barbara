```sql
--Analyse des remises :
--Calculez le total des remises
--(UnitPriceDiscount) appliquées par produit et identifiez celui avec la remise totale la plus élevée.
--(Table : SalesOrderDetail)

select
 det.ProductID,
 sum(det.UnitPriceDiscount) as TotalRemise
from 
atb_adventureworks.saleslt.salesorderdetail det
group by det.productid
order by TotalRemise desc;

select
 det.ProductID,
 sum(det.UnitPriceDiscount) as TotalRemise
from 
atb_adventureworks.saleslt.salesorderdetail det
group by det.productid
order by TotalRemise desc
limit 1;
```
