```sql
--Top 3 clients générant le plus de
--revenus :
--Identifiez les 3 clients ayant généré le
--plus de revenus totaux à travers leurs
--commandes.
--(Tables : SalesOrderHeader, Customer)

select 
concat(cus.Title, ' ', cus.FirstName, ' ', cus.LastName) as Client,
sum(sal.TotalDue) as Revenus
from 
atb_adventureworks.saleslt.customer cus,
atb_adventureworks.saleslt.salesorderheader sal
where
cus.CustomerID = sal.CustomerID
group by
cus.Title, cus.FirstName, cus.LastName
order by
sum(sal.TotalDue) desc
limit 3;
  
```
