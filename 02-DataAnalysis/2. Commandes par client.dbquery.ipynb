```sql
--Commandes par client :
--Affichez le nombre total de
--commandes (SalesOrderID) effectuées
--par chaque client.
--(Tables : SalesOrderHeader, Customer)

select 
concat(cus.Title, ' ', cus.FirstName, ' ', cus.LastName) as Client,
count(sal.SalesOrderID) as NombreCommandes
from 
atb_adventureworks.saleslt.salesorderheader sal, 
atb_adventureworks.saleslt.customer cus
where 
sal.CustomerID = cus.CustomerID
group by
concat(cus.Title, ' ', cus.FirstName, ' ', cus.LastName);
```
