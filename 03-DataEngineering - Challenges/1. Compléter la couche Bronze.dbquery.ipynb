```sql
--Compléter la couche Bronze:
--Récupérez dans Databricks toutes les
--tables de la base de données
--AdventureWorksLT, sauf les tables de
--metadata (i.e. ErrorLog, BuildVersion).

SHOW TABLES IN atb_adventureworks.saleslt;

SHOW TABLES IN barbara_lakehouse.bronze;

-- Génération des requêtes
SELECT
  CONCAT(
    'CREATE OR REPLACE TABLE barbara_lakehouse.bronze.', lower(TABLE_NAME),
    ' USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.', TABLE_NAME, ';'
  ) AS sql_statement
FROM atb_adventureworks.information_schema.tables
 WHERE TABLE_NAME NOT IN ('ErrorLog', 'BuildVersion');

-- Exécution des requêtes
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.address USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.Address;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.customeraddress USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.CustomerAddress;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.salesorderdetail USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.SalesOrderDetail;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.salesorderheader USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.SalesOrderHeader;
--CREATE OR REPLACE TABLE barbara_lakehouse.bronze.database_firewall_rules USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.database_firewall_rules;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.customer USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.Customer;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.productmodel USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.ProductModel;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.vproductmodelcatalogdescription USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.vProductModelCatalogDescription;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.productdescription USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.ProductDescription;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.product USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.Product;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.productmodelproductdescription USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.ProductModelProductDescription;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.vproductanddescription USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.vProductAndDescription;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.productcategory USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.ProductCategory;
CREATE OR REPLACE TABLE barbara_lakehouse.bronze.vgetallcategories USING DELTA AS SELECT * FROM atb_adventureworks.saleslt.vGetAllCategories;
```
